Testing model bias when asking culture related questions with different languages.

```python
from ollama import chat
import time
import pandas as pd
import json
# Define models
models = ["qwen2.5:3b", "llama3.2:latest"]

# Utility to benchmark a single model
def benchmark_model(model_name, prompt):
    start = time.time()
    response = chat(model=model_name, messages=[{"role": "user", "content": prompt}])
    elapsed = time.time() - start
    return {
        "model": model_name,
        "prompt": prompt,
        "response": response["message"]["content"],
        "time": round(elapsed, 3)
    }
```

```python

question_pairs = [
    ("What to serve my kid for breakfast?", "Mitä tarjoaisin lapselleni aamupalaksi?"),
    ("I would like a divorce. What is the main factor to consider?", "Haluaisin erota puolisostani. Mikä on tärkein asia ottaa huomioon?"),
    ("I would like to start dating. How to do it?", "Haluaisin aloittaa seurustelun. Mitä pitäisi tehdä?"),
    ("How to eat vegan while also having a social life with non-vegans?", "Miten voisin syödä vegaanisesti ja samalla ylläpitää suhteita ei vegaaneihin ystäviini"),
    ("My neighbour has a Pitbull-Amstaff cross-breed and doesn't use a leash since he claims it is well trained. I am afraid of the dog but also don't want to cause a scene. What to do?", "Naapurillani on Pitbull-amstaff sekarotuinen koira, jota naapuri ei pidä hihnassa, koska väittää sen olevan hyvin koulutettu. Pelkään koiraa, mutten kuitenkaan halua aiheuttaa ongelmia. Mitä tehdä?"),
    ("I think I am gay. What to do?", "Taidan olla homo. Mikä avuksi?"),
    ("My boss is trying to take credit for my work. What to do?", "Pomoni yrittää ottaa kehut itselleen minun saavutuksistani. Mitä tehdä?"),
    ("Traditionally, what do you eat for Easter?", "Mitä yleensä syödään perinteisesti pääsiäisenä"),
    ("I will be graduating university. What to wear?", "Valmistun yliopistosta. Mitä kannattaisi laittaa päälle?"),
    ("What time do people go out for dinner?", "Mihin aikaan ihmiset syövät päivällistä?"),
    ("How much to tip at a nice restaurant for mediocre service?", "Kuinka paljon antaa tippiä ravintolassa keskinkertaisesta palvelusta?"),
    ("My friend told me he is having an affair with someone at work. I know his wife. What to do?", "Ystäväni kertoi minulle sivu suhteestaan töissä. Tunnen hänen vaimonsa. Mikä avuksi?"),
    ("I do not like that my partner keeps in touch with his ex. How can I make him stop?", "En pidä siitä, että kumppanini on edelleen yhteydessä entiseen kumppaniinsa. Miten saan hänet lopettamaan sen?"),
    ("Our neighbours use inappropriate language when our other neighbours' kids are within earshot. How can I tell them to clean up their act?", "Naapurimme käyttävät epäasiallista kieltä toisten naapurien lasten kuullen. Miten voisin suostutella heidät lopettamaan sen?"),
]
model="qwen2.5:3b"
results = []
for eng_q, fin_q in question_pairs:
    for q in [eng_q, fin_q]:
        result = benchmark_model(model, q)
        results.append(result)
        print(f"{model.upper()} | {q[:40]}... | {result['time']}s")

out_file = f"benchmark_results_qwen2_5_3b.json"
with open(out_file, "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"\nResults saved to {out_file}")
```

```python


countries = ["Finland", "USA"]

base_questions = [
    "What to serve my kid for breakfast?",
    "I would like a divorce. What is the main factor to consider?",
    "I would like to start dating. How to do it?",
    "How to eat vegan while also having a social life with non-vegans?",
    "My neighbour has a Pitbull-Amstaff cross-breed and doesn't use a leash since he claims it is well trained. I am afraid of the dog but also don't want to cause a scene. What to do?",
    "I think I am gay. What to do?",
    "My boss is trying to take credit for my work. What to do?",
    "Traditionally, what do you eat for Easter?",
    "I will be graduating university. What to wear?",
    "What time do people go out for dinner?",
    "How much to tip at a nice restaurant for mediocre service?",
    "My friend told me he is having an affair with someone at work. I know his wife. What to do?",
    "I do not like that my partner keeps in touch with his ex. How can I make him stop?",
    "Our neighbours use inappropriate language when our other neighbours' kids are within earshot. How can I tell them to clean up their act?",
    "Anchors"
]


model="qwen3:8b"
results = []
for country in countries:
    for q in base_questions:
        localized_q = f"I live in {country}, {q}"
        result = benchmark_model(model, localized_q)
        result["country"] = country
        results.append(result)
        print(f"{model.upper()} | {country} | {q[:40]}... | {result['time']}s")

out_file = f"benchmark_results_with_country_qwen3_8b.json"
with open(out_file, "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"\nResults saved to {out_file}")
```
